```python
import pandas as pd
import numpy as np
import csv
```

```python
url = "https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv"
df=pd.read_csv(url)
df
```

```python
print(df.isna().sum())
```

```python
print("Number of duplicates:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
df =df.rename(columns={"Title":"Movie_Title"})
df
```

```python
df["Genre"].unique()
```

```python
df["Genre_Count"] = df["Genre"].str.split(",").str.len()

display(df[["Movie_Title", "Genre", "Genre_Count"]].head(10))
```

```python
print(df[["Revenue (Millions)", "Metascore"]].isna().sum())


```

```python
revenue_mean = df["Revenue (Millions)"].mean()

metascore_mean = df["Metascore"].mean()

print("Revenue Mean:", revenue_mean)
print("Metascore Mean:", metascore_mean)
```

```python
df["Revenue (Millions)"] = df["Revenue (Millions)"].fillna(revenue_mean)

df["Metascore"] = df["Metascore"].fillna(metascore_mean)
```

```python
print(df[["Revenue (Millions)", "Metascore"]].isna().sum())
```

```python
top_5 = df.sort_values(by="Rating", ascending=False).head(10)
top_5
```

```python
revenue_high=df.sort_values(by="Revenue (Millions)",ascending=False).head(1)

revenue_low=df.sort_values(by="Revenue (Millions)",ascending=True).head(1)
revenue_low
```

```python
avg_rating = df.groupby('Director')['Rating'].mean()
print(avg_rating)
```

```python
movies_per_year = df.groupby('Year')['Movie_Title'].count()

print(movies_per_year)
```

```python
result = df[(df['Rating'] > 8) & (df['Revenue (Millions)'] > 100)]
result[['Movie_Title', 'Rating', 'Revenue (Millions)']]
```

```python
result = df.groupby('Director')[['Rating', 'Revenue (Millions)']].mean()

result
```

```python
result = df.groupby('Year').agg({
    'Movie_Title': 'count',
    'Rating': 'mean'
})
result
```
